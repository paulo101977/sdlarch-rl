```python
import ctypes
import numpy as np
import time
import cv2
import gc
import os
import zlib
from sdlarch_rl import make
import pygame
import gzip
from sdlarch_rl.utils.utils import ExcludeButtonsWrapper, RandomStateWrapper

# env = make("SuperStreetFighterIV-3DS", statename="ryu_ken_hardest")
env = make("SuperStreetFighterIV-3DS")
# env = RandomStateWrapper(env, states=['default', 'ryu_ken_easy_south_asia', 'ryu_guile_easy_skyscraper', 'ryu_chunli_easy_asia_south'])
#env = RandomStateWrapper(env, states=['ryu_ken_easy'])
env = RandomStateWrapper(env, states=['ryu_ken_hardest_original'])

count = 0
global initial_state
initial_state = None

pygame.init()

SCREEN_WIDTH = 640
SCREEN_HEIGHT = 480

if os.name == 'nt':
    SCREEN_WIDTH = 768*2
    SCREEN_HEIGHT = 640*2
    
window = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))
clock = pygame.time.Clock()

buttons = env.unwrapped.buttons
to_exclude = ["START", "SELECT", "L2", "R2", "L3", "R3", "A", "X", "Y"]

env = ExcludeButtonsWrapper(env, buttons, to_exclude)

obs, info = env.reset()
player2 = 0
prev_player2 = 0
max_diff = 0
count = 0

while True:
    pygame.event.pump()
    # action = np.zeros(16, dtype=np.uint8)
    action = env.action_space.sample()
    
    obs, rew, done, _, info = env.step(action)

    obs = cv2.resize(obs, (int(SCREEN_WIDTH/10), int(SCREEN_HEIGHT/10)))
    obs = cv2.resize(obs, (SCREEN_WIDTH, SCREEN_HEIGHT))

    # obs = remove_bg(obs)
    
    surface = pygame.surfarray.make_surface(np.transpose(obs, (1, 0, 2)))

    print(info)
    player1 = info.get('player1', 0)
    player2 = info.get('player2', 0)
    

    window.blit(surface, (0, 0))
    pygame.display.update()

    if player2 == 0 or player1 == 0:
        obs, info = env.reset()

    # clock.tick(60)

print("diff, count: ", max_diff, count)
    
```
